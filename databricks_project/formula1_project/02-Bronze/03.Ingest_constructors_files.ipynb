```python
sources_path = f"{landing_folder_path}/constructors.json"
table_name = f"{catalog_name}.{bronze_schema}.constructors"
```

```python
from pyspark.sql.types import StructType,StructField,StringType,IntegerType,DoubleType


schema=StructType([
    StructField('constructorId',StringType()),
    StructField('name',StringType()),
    StructField('nationality',StringType()),
    StructField('url',StringType())
    ]
)
```

```python
constructors_df=(
    spark.read.format('json')
    #.option('header',True)
    #.option('inferSchema',True)
    .option('mode','FAILFAST')
    .schema(schema)
    .load(sources_path))
```

### Step 2 - Ingest time stamp and file path

```python
from pyspark.sql.functions import *
```

### Step3- Write data to Bronze delta table

```python
constructors_final_df.write.mode('overwrite').format('delta').saveAsTable(table_name)
```
