```sql

create or replace view formula1.gold.driver_standings as
SELECT DISTINCT
       a.season,
       a.driver_id,
       b.driver_name,
       b.nationality,
       count(*) as number_participations,
       sum(a.points) as Total_points,
       rank() over(partition by a.season order by sum(a.points) desc,count_if(a.is_win)
        , count_if(a.is_podium) desc) as standing_position,

       count_if(a.is_win) as number_wins,
       count_if(a.is_podium) as number_podiums

FROM formula1.gold.results_fact as a
join formula1.gold.dim_drivers  as b
on a.driver_id = b.driver_id

group BY
       a.season,
       a.driver_id,
       b.driver_name,
       b.nationality

;
```

```sql
select * from formula1.gold.driver_standings a where a.season = 2023;
```
