# 01.Setup the project environment for formula 1
- Create new extrenal location for your container formula 1
- Create new catalog for formula1 project

External Location

```python
%sql
create external location if not exists databricks_ext_9860_formula1_incr
url 'abfss://formula1-incr@databricksextdl9860.dfs.core.windows.net/'
with (storage credential `databricks-course-sc`)
comment 'External location for the formula 1 incr container';
```

Access cloud containers storage

```python
 %fs ls 'abfss://formula1-incr@databricksextdl9860.dfs.core.windows.net/landing/'
```

New Catalog for Formula1

```python
%sql
show catalogs;
```

```python
%sql
create catalog if not exists formula1_incr
managed location 'abfss://formula1-incr@databricksextdl9860.dfs.core.windows.net/'
comment 'this is a catalog for the formula 1 database'
;
```

Create Schema/Database for all bronze gold silver layer

```python
%sql
create schema if not exists formula1_incr.landing;
create schema if not exists formula1_incr.bronze
    managed location 'abfss://formula1-incr@databricksextdl9860.dfs.core.windows.net/bronze';

create schema if not exists formula1_incr.silver
    managed location 'abfss://formula1-incr@databricksextdl9860.dfs.core.windows.net/silver';

create schema if not exists formula1_incr.gold
    managed location 'abfss://formula1-incr@databricksextdl9860.dfs.core.windows.net/gold'
```
